```python
activities=[(480,600,5),(510,570,3),(630,690,4),(480,720,20),(630,690,1)]
cranky=[540,660]
```

```python
def first_ok(i,activities,cranky):
    for t in cranky:
        if t<activities[i][0]:
            return False
    return True
```

```python
for j in range(len(activities)):
    print(j,first_ok(j,activities,cranky))
```

```python
def compatible(j,i,activities,cranky):
    if activities[j][1]>activities[i][0]:
        return False
    for t in cranky:
        if t>activities[j][1] and t<activities[i][0]:
            return False
    return True
```

```python
for j in range(len(activities)):
    for  i in range(j+1,len(activities)):
        print(j,"-",i,compatible(j,i,activities,cranky))
```
