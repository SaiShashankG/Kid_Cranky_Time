```python
activities=[(480,600,5),(510,570,3),(630,690,4),(480,720,20),(690,750,1)]
cranky=[540,660]
n=len(activities)
```

```python
def first_ok(i,activities,cranky):
    for t in cranky:
        if t<activities[i][0]:
            return False
    return True
```

```python
for j in range(len(activities)):
    print(j,first_ok(j,activities,cranky))
```

```python
def compatible(j,i,activities,cranky):
    if activities[j][1]>activities[i][0]:
        return False
    for t in cranky:
        if t>activities[j][1] and t<activities[i][0]:
            return False
    return True
```

```python
for j in range(len(activities)):
    for  i in range(j+1,len(activities)):
        print(j,"-",i,compatible(j,i,activities,cranky))
```

```python
def last_ok(i,activities,cranky):
    for t in cranky:
        if t>activities[i][1]:
            return False
    return True
```

```python
for j in range(len(activities)):
    print(j,last_ok(j,activities,cranky))
```

```python
def schedule_activities(activities,cranky):
    if len(cranky)==0:
        return 0
    
    activities=sorted(activities, key=lambda a: a[1])
    n=len(activities)
    INF=float('inf')
    f=[INF]*n

    ans=INF

    for i in range(n):
        if first_ok(i,activities,cranky):
            best= 0
        else:
            best=INF
            
        for j in range(i):
            if compatible(j,i,activities,cranky):
                best=min(best,f[j])
        f[i]=activities[i][2]+best

        if last_ok(i,activities,cranky):
            ans=min(ans,f[i])
            
    return ans
```

```python
print(schedule_activities(activities,cranky))
```

```python
print(schedule_activities([(8, 9, 2), (10, 11, 3), (12, 13, 4)],[8.5, 10.5, 12.5]))
```

```python
print(schedule_activities([(8, 9, 2), (10, 11, 3), (12, 13, 4)],[]))
```

```python
def no_overlap(chosen):
    chosen=sorted(chosen,key= lambda a:a[1])
    for i in range(1,len(chosen)):
        if chosen[i-1][1]>chosen[i][0]:
            return False
    return True
```

```python
no_overlap([(8, 9, 1), (10, 11, 1)])
```

```python
no_overlap([(8, 10, 1), (9, 11, 1)])
```

```python
def all_covered(chosen,cranky):
    for i in cranky:
        covered=False
        for j in range(len(chosen)):
            if i>=chosen[j][0] and i<=chosen[j][1]:
                covered=True
                break
        if not covered:
            return False
    return True
```

```python
all_covered([(8, 10, 1)], [9])
```

```python
all_covered([(8, 10, 1)], [9, 11])
```

```python
all_covered([], [9])
```

```python
def brute_force(activities,cranky):
    INF=float('inf')
    best=INF
    n=len(activities)

    for k in range(n+1):
        for subset in combinations(activities,k):
            if no_overlap(subset) and all_covered(subset,cranky):
                cost=sum(s[2] for s in subset)
                best=min(best,cost)
    return best
```

```python
print(brute_force(activities,cranky))
print(brute_force([(8, 9, 2), (10, 11, 3), (12, 13, 4)],[8.5, 10.5, 12.5]))
print(brute_force([(8, 9, 2), (10, 11, 3), (12, 13, 4)],[]))
```

```python
## Comparintg brute force with our algorithm
import random

def random_test(trials=2000):
    for _ in range(trials):
        n = random.randint(1, 8)
        activities = []
        for _ in range(n):
            s = random.randint(0, 20)
            e = random.randint(s + 1, 24)
            fee = random.randint(1, 10)
            activities.append((s, e, fee))
        cranky = [random.randint(0, 24) for _ in range(random.randint(1, 4))]
        print(activities,cranky)
        if schedule_activities(activities,cranky)!=brute_force(activities,cranky):
            return False
        return True
```

```python
print(random_test())
```

In the `schedule_activities.py` \
**Time Complexity Anlaysis** 

let len(cranky)=m and len(activities)=n
- first\_ok --> $O(m)$ 
- compatibility --> $O(m) $
- last_ok --> $O(m) $

Then, for over all algorithm it takes $O(n^2 + nm)$

In the `brute_force.py` \
**Time Complexity Analysis**

- no_overlap --> $O(klogk)$ , k - length of subset
- all_covered --> $O(mk)$

Then, for over all brute force algorithm it takes $O(mn^2 2^n)$
